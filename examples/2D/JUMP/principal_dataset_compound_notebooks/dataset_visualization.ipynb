# JUMP ORF Dataset Visualization : High-quality channel visualization for multi-channel microscopy data


```python
# =============================================================================
# JUMP ORF Dataset Visualization for Publication
# Modified version with legend and zoom capabilities
# =============================================================================

import os
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
from matplotlib.gridspec import GridSpec
from matplotlib.patches import Rectangle
import polars as pl
import requests
import seaborn as sns
from jump_portrait.fetch import get_item_location_info, get_jump_image
# Check if image metadata contains pixel size

# For TIFF files, metadata would be accessible via:
# tifffile.TiffFile(path).pages[0].tags
# Look for 'XResolution', 'YResolution', or custom metadata

# Set publication-quality plotting parameters
plt.rcParams.update({
    'font.size': 10,
    'font.family': 'sans-serif',
    'font.sans-serif': ['Arial'],
    'axes.labelsize': 10,
    'axes.titlesize': 11,
    'xtick.labelsize': 9,
    'ytick.labelsize': 9,
    'legend.fontsize': 9,
    'figure.dpi': 300,
    'savefig.dpi': 300,
    'savefig.bbox': 'tight',
    'savefig.pad_inches': 0.05
})

# =============================================================================
# 1. Load ORF Profile Data
# =============================================================================

print("=== Loading JUMP ORF Dataset ===")
INDEX_FILE = "https://raw.githubusercontent.com/jump-cellpainting/datasets/v0.11.0/manifests/profile_index.json"
response = requests.get(INDEX_FILE)
profile_index = response.json()

orf_url = None
for dataset in profile_index:
    if dataset['subset'] == 'orf':
        orf_url = dataset['url']
        break

profiles = pl.scan_parquet(orf_url)
print("✓ ORF profiles loaded")

# =============================================================================
# 2. Load ORF Metadata (JCP2022 -> Gene Symbol Mapping)
# =============================================================================

print("\n=== Loading ORF Metadata ===")
metadata_url = "https://raw.githubusercontent.com/jump-cellpainting/datasets/main/metadata/orf.csv.gz"
response = requests.get(metadata_url)

import gzip
from io import StringIO
import pandas as pd
content = gzip.decompress(response.content).decode('utf-8')
orf_metadata_df = pd.read_csv(StringIO(content))

# Create JCP2022 -> Gene Symbol mapping
jcp_to_gene = dict(zip(orf_metadata_df['Metadata_JCP2022'], 
                       orf_metadata_df['Metadata_Symbol']))
print(f"✓ Loaded mappings for {len(jcp_to_gene)} ORFs")

# =============================================================================
# 3. Select Sample ORFs for Visualization
# =============================================================================

selected_orfs = [
    'JCP2022_904947', 'JCP2022_906496', 'JCP2022_904829', 'JCP2022_900076',
    'JCP2022_904928', 'JCP2022_905988', 'JCP2022_903771', 'JCP2022_906823'
]

print(f"\n=== Selected {len(selected_orfs)} ORFs for visualization ===")

# =============================================================================
# 4. Channel Configuration
# =============================================================================

CHANNELS = ["DNA", "RNA", "ER", "AGP", "Mito"]

# Channel colors based on fluorophore emission
CHANNEL_COLORS = {
    "DNA": "#4A90E2",   # Blue
    "RNA": "#F5A623",   # Yellow
    "ER": "#7ED321",    # Green
    "AGP": "#FF6B6B",   # Red-orange
    "Mito": "#D0021B"   # Deep red
}

# Reagent information for legend
CHANNEL_REAGENTS = {
    "DNA": "Hoechst 33342",
    "RNA": "SYTO 14",
    "ER": "Concanavalin A / Alexa 488",
    "AGP": "Phalloidin / WGA",
    "Mito": "MitoTracker Deep Red"
}

CHANNEL_TARGETS = {
    "DNA": "DNA (Nucleus)",
    "RNA": "RNA (Cytoplasmic RNA, Nucleoli)",
    "ER": "ER (Endoplasmic Reticulum)",
    "AGP": "AGP (Actin, Golgi, Plasma Membrane)",
    "Mito": "Mitochondria"
}

# =============================================================================
# 5. Core Functions
# =============================================================================

def normalize_image(img, percentile_clip=True, lower=1, upper=99):
    """Normalize image intensity to [0, 1] with percentile clipping"""
    img = img.astype(np.float32)
    
    if percentile_clip:
        p_low = np.percentile(img, lower)
        p_high = np.percentile(img, upper)
        img = np.clip(img, p_low, p_high)
    
    img_min, img_max = img.min(), img.max()
    if img_max > img_min:
        img = (img - img_min) / (img_max - img_min)
    
    return img


def create_colormap(color_hex):
    """Create colormap from black to specified color"""
    color_rgb = mcolors.hex2color(color_hex)
    return mcolors.LinearSegmentedColormap.from_list(
        'custom', ['black', color_rgb]
    )


def create_rgb_composite(channel_images):
    """Create RGB composite from 5-channel data"""
    shape = channel_images['DNA'].shape
    composite = np.zeros((*shape, 3), dtype=np.float32)
    
    # Blue channel: DNA
    composite[..., 2] = channel_images['DNA']
    
    # Green channel: ER (primary) + RNA contribution
    composite[..., 1] = 0.7 * channel_images['ER'] + 0.3 * channel_images['RNA']
    
    # Red channel: Mito (primary) + AGP contribution  
    composite[..., 0] = 0.6 * channel_images['Mito'] + 0.4 * channel_images['AGP']
    
    composite = np.clip(composite, 0, 1)
    
    return composite


def add_scalebar(ax, pixel_size_um=0.65, length_um=20, 
                 location='lower right', color='white'):
    """Add scalebar to image"""
    from matplotlib.offsetbox import AnchoredOffsetbox, AuxTransformBox
    
    length_px = length_um / pixel_size_um
    
    box = AuxTransformBox(ax.transData)
    rect = Rectangle((0, 0), length_px, 3, fc=color, ec=color)
    box.add_artist(rect)
    
    loc_dict = {
        'lower right': 4, 'lower left': 3,
        'upper right': 1, 'upper left': 2
    }
    
    anchored_box = AnchoredOffsetbox(
        loc=loc_dict[location],
        child=box,
        frameon=False,
        pad=0.5,
        bbox_to_anchor=(0.98, 0.05) if 'right' in location else (0.02, 0.05),
        bbox_transform=ax.transAxes
    )
    
    ax.add_artist(anchored_box)
    
    ax.text(
        0.98 if 'right' in location else 0.02,
        0.12,
        f'{length_um} μm',
        transform=ax.transAxes,
        color=color,
        fontsize=8,
        ha='right' if 'right' in location else 'left',
        va='bottom',
        weight='bold'
    )


def create_channel_legend(fig, bbox=(0.92, 0.3, 0.15, 0.4)):
    """
    Create channel legend showing color coding and reagents
    
    Parameters
    ----------
    fig : matplotlib.figure.Figure
        Figure to add legend to
    bbox : tuple
        (left, bottom, width, height) in figure coordinates
    """
    from matplotlib.patches import Patch
    
    # Create legend elements
    legend_elements = []
    for channel in CHANNELS:
        label = f"{CHANNEL_TARGETS[channel]}\n{CHANNEL_REAGENTS[channel]}"
        legend_elements.append(
            Patch(facecolor=CHANNEL_COLORS[channel], 
                  edgecolor='black', 
                  label=label,
                  linewidth=0.5)
        )
    
    # Add legend to figure
    legend_ax = fig.add_axes(bbox)
    legend_ax.axis('off')
    
    legend = legend_ax.legend(
        handles=legend_elements,
        loc='center',
        frameon=True,
        fancybox=False,
        shadow=False,
        title='Channel Information',
        title_fontsize=9,
        fontsize=7,
        handlelength=1.5,
        handleheight=1.5,
        labelspacing=1.2
    )
    
    legend.get_frame().set_linewidth(1.0)
    legend.get_frame().set_edgecolor('black')
    legend.get_title().set_weight('bold')
    
    return legend_ax


# =============================================================================
# 6. Full Field of View Figure
# =============================================================================

def create_full_fov_figure(orf_id, jcp_to_gene, image_idx=0, save_path=None):
    """
    Create full field-of-view figure with all channels and legend.
    
    Layout: 2 rows x 3 columns + legend on right
    """
    
    gene_symbol = jcp_to_gene.get(orf_id, "Unknown")
    
    # Retrieve image location
    gene_info = get_item_location_info(gene_symbol)
    orf_samples = gene_info.filter(pl.col("Metadata_PlateType") == "ORF")
    
    if orf_samples.shape[0] == 0:
        print(f"No ORF images found for {orf_id}")
        return None
    
    row = orf_samples.row(image_idx)
    source = row[gene_info.columns.index("Metadata_Source")]
    batch = row[gene_info.columns.index("Metadata_Batch")]
    plate = row[gene_info.columns.index("Metadata_Plate")]
    well = row[gene_info.columns.index("Metadata_Well")]
    site = row[gene_info.columns.index("Metadata_Site")]
    
    print(f"Loading: {gene_symbol} ({orf_id})")
    print(f"  Location: {source}/{batch}/{plate}/{well}/site_{site}")
    
    # Load all channels
    channel_images = {}
    channel_images_norm = {}
    
    for channel in CHANNELS:
        img = get_jump_image(source, batch, plate, well, channel, site, None)
        channel_images[channel] = img
        channel_images_norm[channel] = normalize_image(img, percentile_clip=True)
    
    composite = create_rgb_composite(channel_images_norm)
    
    # Create figure with extra space for legend
    fig = plt.figure(figsize=(14, 8))
    
    # Main grid for images (left 85% of figure)
    gs = GridSpec(2, 3, figure=fig, 
                  left=0.05, right=0.85, top=0.92, bottom=0.08,
                  hspace=0.60, wspace=0.15)
    
    # Plot individual channels
    for idx, channel in enumerate(CHANNELS):
        row_idx = idx // 3
        col_idx = idx % 3
        ax = fig.add_subplot(gs[row_idx, col_idx])
        
        cmap = create_colormap(CHANNEL_COLORS[channel])
        ax.imshow(channel_images_norm[channel], cmap=cmap, interpolation='none')
        
        ax.set_title(CHANNEL_TARGETS[channel], fontweight='bold', pad=8)
        ax.axis('off')
        

    
    # Plot composite
    ax_composite = fig.add_subplot(gs[1, 2])
    ax_composite.imshow(composite, interpolation='none')
    ax_composite.set_title('RGB Composite', fontweight='bold', pad=8)
    ax_composite.axis('off')

    
    # Add channel legend
    create_channel_legend(fig, bbox=(0.87, 0.15, 0.12, 0.7))
    
    # Figure title
    fig.suptitle(f'{gene_symbol} ORF Overexpression \n Plate {plate} | Well {well} | Site {site}',
             fontsize=12, fontweight='bold', y=1)  # Changed y from 0.97 to 0.96
    
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight', facecolor='white')
        print(f"✓ Saved: {save_path}")
    
    return fig, channel_images_norm, composite


# =============================================================================
# 7. Zoomed/Cropped Figure
# =============================================================================

def create_zoomed_figure(orf_id, jcp_to_gene, image_idx=0, 
                         crop_center=None, crop_size=300, save_path=None):
    """
    Create zoomed figure showing cropped regions from all channels.
    
    Parameters
    ----------
    orf_id : str
        JCP2022 ID
    jcp_to_gene : dict
        Mapping dictionary
    image_idx : int
        Which replicate image to use
    crop_center : tuple or None
        (y, x) coordinates for crop center. If None, uses image center
    crop_size : int
        Size of cropped region in pixels
    save_path : str or None
        Path to save figure
    """
    
    gene_symbol = jcp_to_gene.get(orf_id, "Unknown")
    
    # Retrieve image location
    gene_info = get_item_location_info(gene_symbol)
    orf_samples = gene_info.filter(pl.col("Metadata_PlateType") == "ORF")
    
    if orf_samples.shape[0] == 0:
        print(f"No ORF images found for {orf_id}")
        return None
    
    row = orf_samples.row(image_idx)
    source = row[gene_info.columns.index("Metadata_Source")]
    batch = row[gene_info.columns.index("Metadata_Batch")]
    plate = row[gene_info.columns.index("Metadata_Plate")]
    well = row[gene_info.columns.index("Metadata_Well")]
    site = row[gene_info.columns.index("Metadata_Site")]
    
    print(f"Loading zoomed view: {gene_symbol} ({orf_id})")
    
    # Load all channels
    channel_images_norm = {}
    
    for channel in CHANNELS:
        img = get_jump_image(source, batch, plate, well, channel, site, None)
        channel_images_norm[channel] = normalize_image(img, percentile_clip=True)
    
    composite = create_rgb_composite(channel_images_norm)
    
    # Determine crop region
    img_shape = channel_images_norm['DNA'].shape
    if crop_center is None:
        crop_center = (img_shape[0] // 2, img_shape[1] // 2)
    
    half_size = crop_size // 2
    y_start = max(0, crop_center[0] - half_size)
    y_end = min(img_shape[0], crop_center[0] + half_size)
    x_start = max(0, crop_center[1] - half_size)
    x_end = min(img_shape[1], crop_center[1] + half_size)
    
    # Crop all images
    channel_images_cropped = {}
    for channel in CHANNELS:
        channel_images_cropped[channel] = channel_images_norm[channel][y_start:y_end, x_start:x_end]
    
    composite_cropped = composite[y_start:y_end, x_start:x_end, :]
    
    # Create figure
    fig = plt.figure(figsize=(14, 8))
    
    gs = GridSpec(2, 3, figure=fig, 
              left=0.05, right=0.85, top=0.88, bottom=0.08,  # Changed top from 0.92 to 0.88
              hspace=0.40, wspace=0.15)  # Changed hspace from 0.25 to 0.30
    
    # Plot cropped channels
    for idx, channel in enumerate(CHANNELS):
        row_idx = idx // 3
        col_idx = idx % 3
        ax = fig.add_subplot(gs[row_idx, col_idx])
        
        cmap = create_colormap(CHANNEL_COLORS[channel])
        ax.imshow(channel_images_cropped[channel], cmap=cmap, interpolation='none')
        
        ax.set_title(CHANNEL_TARGETS[channel], fontweight='bold', pad=8)
        ax.axis('off')
        


    
    # Plot composite
    ax_composite = fig.add_subplot(gs[1, 2])
    ax_composite.imshow(composite_cropped, interpolation='none')
    ax_composite.set_title('RGB Composite', fontweight='bold', pad=8)
    ax_composite.axis('off')

    
    # Add channel legend
    create_channel_legend(fig, bbox=(0.87, 0.15, 0.12, 0.7))
    
    # Figure title
    crop_um = crop_size * 0.65
    fig.suptitle(f'{gene_symbol} ORF (Overexpression) \n Plate {plate} | Well {well} | Site {site}',
                 fontsize=12, fontweight='bold', y=0.96)
    
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight', facecolor='white')
        print(f"✓ Saved: {save_path}")
    
    return fig, channel_images_cropped, composite_cropped


# =============================================================================
# 8. Execute Visualizations
# =============================================================================

if __name__ == "__main__":
    
    os.makedirs("orf_visualizations", exist_ok=True)
    
    # Full field of view
    print("\n=== Creating full field-of-view visualization ===")
    fig_full, channels_norm, composite = create_full_fov_figure(
        selected_orfs[0],
        jcp_to_gene,
        save_path="orf_visualizations/fig_full_fov.png"
    )
    plt.show()
    
    # Zoomed view - center crop
    print("\n=== Creating zoomed visualization (center) ===")
    fig_zoom, channels_crop, composite_crop = create_zoomed_figure(
        selected_orfs[0],
        jcp_to_gene,
        crop_size=300,  # 300 pixels = ~195 μm
        save_path="orf_visualizations/fig_zoomed_center.png"
    )
    plt.show()
    
    # Zoomed view - custom location (e.g., interesting cellular feature)
    print("\n=== Creating zoomed visualization (custom location) ===")
    fig_zoom2, _, _ = create_zoomed_figure(
        selected_orfs[0],
        jcp_to_gene,
        crop_center=(512, 512),  # Specify y, x coordinates
        crop_size=400,
        save_path="orf_visualizations/fig_zoomed_custom.png"
    )
    plt.show()
    
    print("\n✓ All visualizations complete")
```
