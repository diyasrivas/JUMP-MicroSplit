## Load the checkpoint

```python
# List all available checkpoints
def list_available_model_checkpoints():
    """List all model checkpoints available in the checkpoints and pretrained_checkpoints folders"""
    from pathlib import Path
    
    # Regular checkpoints (trained models)
    ckpt_folder = Path('/home/diya.srivastava/Desktop/repos/JUMP-MicroSplit/examples/2D/JUMP/checkpoints/pilot_5_channels/best-epoch=3.ckpt')
    ckpt_files = list(ckpt_folder.glob("**/*.ckpt")) if ckpt_folder.exists() else []
    
    # Pretrained checkpoints
    pretrained_ckpt_folder = Path("./pretrained_checkpoints")
    pretrained_ckpt_files = list(pretrained_ckpt_folder.glob("**/*.ckpt")) if pretrained_ckpt_folder.exists() else []
    
    print("Available model checkpoints:")
    print("==========================")
    
    if ckpt_files:
        print("\nTrained model checkpoints:")
        for file in ckpt_files:
            print(f"- {file}")
    else:
        print("\nNo trained model checkpoints found.")
        
    if pretrained_ckpt_files:
        print("\nPretrained model checkpoints:")
        for file in pretrained_ckpt_files:
            print(f"- {file}")
    else:
        print("\nNo pretrained model checkpoints found.")
    
    return ckpt_files, pretrained_ckpt_files

ckpt_files, pretrained_ckpt_files = list_available_model_checkpoints()

# Select the most recent checkpoint by default, or specify a path
selected_ckpt = None
if ckpt_files:
    selected_ckpt = str(ckpt_files[0])
elif pretrained_ckpt_files:
    selected_ckpt = str(pretrained_ckpt_files[0])

# Option to manually specify a checkpoint path
user_selected_ckpt = '/home/diya.srivastava/Desktop/repos/JUMP-MicroSplit/examples/2D/JUMP/checkpoints/pilot_5_channels/best-epoch=3.ckpt'
if user_selected_ckpt:
    selected_ckpt = user_selected_ckpt

if not selected_ckpt:
    raise ValueError("No model checkpoint found! Please train a model first or provide a pretrained checkpoint.")

print(f"Selected checkpoint: {selected_ckpt}")
```

## Now let's Make Predictions on Test Data

```python
reduce_data = False
if reduce_data:
    print("Using REDUCED evaluation data for quick testing!")
    dset.reduce_data([0])
else:
    print('Using the full set of evaluation data!')
    print(f'(More specifically, I will use {dset.get_num_frames()} frames for evaluations.)')

from pathlib import Path
import numpy as np
import pandas as pd
import copy  # Add this import

dset._fpath = Path(dset._fpath)

# Load metadata to cross-reference image IDs
try:
    metadata = pd.read_csv("pilot_dataset_expt3/metadata.csv")
    print("Loaded metadata for cross-referencing...")
except:
    print("Could not load metadata.csv - will show indices only")
    metadata = None

# Get the total number of test images
total_test_images = dset.get_num_frames()
print(f"Processing {total_test_images} test images one at a time...")

# Initialize lists to store all predictions and metadata
all_stitched_predictions = []
all_norm_stitched_predictions = []
all_stitched_stds = []
test_metadata = []  # Move this here to collect during the loop

# Store original dataset for restoration
original_dset = dset

# Process each image individually
for i in range(total_test_images):
    print(f"\n=== Processing test image {i+1}/{total_test_images} ===")
    
    # Try to get the actual image_id from metadata
    if metadata is not None:
        # Assuming test images are the last 10% of the dataset
        test_start_idx = int(0.9 * len(metadata))
        actual_image_id = test_start_idx + i
        if actual_image_id < len(metadata):
            img_info = metadata.iloc[actual_image_id]
            print(f"  Original image_id: {img_info['image_id']}")
            print(f"  Plate: {img_info['plate']}")
            print(f"  Well: {img_info['well']}, Site: {img_info['site']}")
            
            # Collect metadata during the loop
            test_metadata.append({
                'test_index': i,
                'original_image_id': img_info['image_id'],
                'batch': img_info['batch'],
                'plate': img_info['plate'],
                'well': img_info['well'],
                'site': img_info['site'],
                'prediction_files': [f"test_pred_frame{i}_{ch}.tif" for ch in TARGET_CHANNEL_LIST],
                'target_files': [f"target_frame{i}_{ch}.tif" for ch in TARGET_CHANNEL_LIST],
                'input_file': f"input_frame{i}_combined.tif"
            })
        else:
            print(f"  Test index: {i} (metadata lookup failed)")
    else:
        print(f"  Test index: {i}")
    
    # Create a proper copy of the dataset for this iteration
    temp_dset = copy.deepcopy(original_dset)  # Use deepcopy for safety
    temp_dset.reduce_data([i])  # Keep only image at index i
    
    print(f"  Processing single image (shape will be processed in tiles)...")
    
    # Generate predictions for this single image
    single_prediction, single_norm_prediction, single_std = get_unnormalized_predictions(
        model=model,
        dset=temp_dset, 
        target_channel_list=TARGET_CHANNEL_LIST,
        mmse_count=50,
        num_workers=4,
        batch_size=32
    )
    
    # Store the results
    all_stitched_predictions.append(single_prediction[0])  # [0] because it's a single image
    all_norm_stitched_predictions.append(single_norm_prediction[0])
    all_stitched_stds.append(single_std[0])
    
    print(f"  ✓ Completed image {i+1}/{total_test_images}")
    print(f"    Prediction shape: {single_prediction[0].shape}")

# Convert lists to numpy arrays
print(f"\n=== Combining all predictions ===")
stitched_predictions = np.array(all_stitched_predictions)
norm_stitched_predictions = np.array(all_norm_stitched_predictions)
stitched_stds = np.array(all_stitched_stds)

print(f"All {total_test_images} images processed successfully!")
print(f"Final prediction shape: {stitched_predictions.shape}")
print(f"Expected shape: ({total_test_images}, 1080, 1080, {len(TARGET_CHANNEL_LIST)})")

# Get target and input data for evaluation (restore original dataset)
dset = original_dset  # Restore full dataset
tar = get_target(dset)
inp = get_input(dset)

print(f"Target shape: {tar.shape}")
print(f"Input shape: {inp.shape}")
print(f"Collected metadata for {len(test_metadata)} test images")
```

## Quantitative Evaluation

```python
# Create output directory
channel_names = "_".join(ch for ch in TARGET_CHANNEL_LIST)
save_dir = f"pilot_expt3_test2_{channel_names}"
os.makedirs(save_dir, exist_ok=True)
import pandas as pd

# Calculate metrics
psnr_results = []
ssim_results = []
for ch_idx in range(tar.shape[-1]):
    # Get PSNR for this channel
    psnr_val = avg_range_inv_psnr(
        [tar[i,...,ch_idx] for i in range(tar.shape[0])], 
        [stitched_predictions[i,...,ch_idx] for i in range(stitched_predictions.shape[0])]
    )
    psnr_results.append(psnr_val)
    
    # Get SSIM for this channel
    ssim_vals = []
    for i in range(tar.shape[0]):
        ssim_vals.append(structural_similarity(
            tar[i,...,ch_idx], 
            stitched_predictions[i,...,ch_idx],
            data_range=tar[i,...,ch_idx].max() - tar[i,...,ch_idx].min()
        ))
    ssim_results.append((np.mean(ssim_vals), np.std(ssim_vals)/np.sqrt(len(ssim_vals))))

# Print results 
print("\nQuantitative Evaluation:")
print("=======================")
for i, channel in enumerate(TARGET_CHANNEL_LIST):
    print(f"Channel {i+1} ({channel}):")
    print(f"  PSNR: {psnr_results[i][0]:.2f} ± {psnr_results[i][1]:.3f}")
    print(f"  SSIM: {ssim_results[i][0]:.4f} ± {ssim_results[i][1]:.4f}")

# Save results to CSV
metrics_data = []
for i, channel in enumerate(TARGET_CHANNEL_LIST):
    metrics_data.append({
        'Channel': channel,
        'PSNR_Mean': psnr_results[i][0],
        'PSNR_Std': psnr_results[i][1],
        'SSIM_Mean': ssim_results[i][0],
        'SSIM_Std': ssim_results[i][1]
    })
metrics_df = pd.DataFrame(metrics_data)
metrics_csv_path = os.path.join(save_dir, "quantitative_metrics.csv")
metrics_df.to_csv(metrics_csv_path, index=False)
print(f"\nQuantitative metrics saved to: {metrics_csv_path}")

```

```python
# Save predictions
for i in range(len(stitched_predictions)):
    for j, channel in enumerate(TARGET_CHANNEL_LIST):
        # Save prediction
        pred_filename = f"test_pred_frame{i}_{channel}.tif"
        tifffile.imwrite(
            os.path.join(save_dir, pred_filename), 
            stitched_predictions[i, ..., j].astype(np.float32)
        )
        
        # Save target for reference
        target_filename = f"target_frame{i}_{channel}.tif"
        tifffile.imwrite(
            os.path.join(save_dir, target_filename), 
            tar[i, ..., j].astype(np.float32)
        )
        
        # Save input image for reference
        input_filename = f"input_frame{i}_combined.tif"
        tifffile.imwrite(
            os.path.join(save_dir, input_filename), 
            inp[i].astype(np.float32)
        )

# SAVE TEST METADATA TO CSV
if test_metadata:
    test_metadata_df = pd.DataFrame(test_metadata)
    test_metadata_csv = os.path.join(save_dir, "test_images_metadata.csv")
    test_metadata_df.to_csv(test_metadata_csv, index=False)
    print(f"Test image metadata saved to: {test_metadata_csv}")
    print(f"Metadata contains {len(test_metadata)} test images with plate/well/site information")
else:
    print("No test metadata collected - metadata.csv may not have been found")

print(f"\nAll predictions and metrics saved to directory: {save_dir}")
print(f"Total files created:")
print(f"  - {len(stitched_predictions) * len(TARGET_CHANNEL_LIST)} prediction files")
print(f"  - {len(stitched_predictions) * len(TARGET_CHANNEL_LIST)} target files") 
print(f"  - {len(stitched_predictions)} input files")
print(f"  - 1 quantitative metrics CSV")
print(f"  - 1 test metadata CSV")
print(f"  = {len(stitched_predictions) * len(TARGET_CHANNEL_LIST) * 2 + len(stitched_predictions) + 2} files total")
print("\nPrediction process complete!")
```
